# 01_bronze_ingestion


Ingest raw retail order data from a CSV file hosted on GitHub 
and store it as a Delta table in the Bronze layer.

## Table of Contents
1. Import libraries and define the data source
2. Load CSV data into a Spark DataFrame
3. Save data as a Bronze Delta table

### Import libraries and define the data source

```python
# 1. Import libraries and define the data source
import pandas as pd

url = (
    "https://raw.githubusercontent.com/"
    "Den24data/databricks-data-pipeline/"
    "main/syntetiske_butiksordrer_2026.csv"
)
```

### Load CSV data into a Spark DataFrame

```python
# 2. Load CSV data into a Spark DataFrame
pdf = pd.read_csv(url)

df = spark.createDataFrame(pdf)

# Preview the raw data
display(df.limit(10))
```

### Save data as a Bronze Delta table

```python
# 3. Save data as a Bronze Delta table
df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "dataengineering_bronze.raw_data.butiksordrer"
    )
```
